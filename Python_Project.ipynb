```python
print("Missing Values:")
print(df.isnull().sum())

print(50*"_")

total_missing = df.isnull().sum().sum()
print("Total Missing Values:", total_missing)

print(50*"_")

print("Number of Duplicate Rows:", df.duplicated().sum())

print(50*"_")

df[df.duplicated()]

print(50*"_")

if df.isnull().sum().sum() > 0:
    df = df.dropna()
    print("Missing values handled.")
else:
    print("No missing values found.")

if df.duplicated().sum() > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")
```

The dataset contains 0 missing values and 0 duplicate rows. No data removal was required.

```python
df_clean = df.copy()

df_clean['Spend'] = pd.to_numeric(
    df_clean['Spend'], errors='coerce' )

print("Spend data type:", df_clean['Spend'].dtype)
print("Invalid values:", df_clean['Spend'].isnull().sum())
```

```python
df_clean.info()
```

```python
print("Dataset Shape:", df_clean.shape)
```

The Spend column was converted to a numeric data type. The 44 invalid values were converted to NaN, while all 1,000 rows were preserved for further analysis.

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df_clean, x='Gender')

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.show()
```

**City Distribution**

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df_clean, x='City')

plt.title('Customer Distribution by City')
plt.xlabel('City')
plt.ylabel('Number of Customers')

plt.xticks(rotation=15)
plt.show()
```

**Food Item Distribution**

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df_clean, x='Item')

plt.title('Food Item Distribution')
plt.xlabel('Food Item')
plt.ylabel('Number of Customers')

plt.xticks(rotation=30)
plt.show()
```

**Eating Frequency Distribution**

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df_clean, x='Frequency')

plt.title('Eating Frequency Distribution')
plt.xlabel('Frequency')
plt.ylabel('Number of Customers')

plt.xticks(rotation=30)
plt.show()
```

**Spending Distribution**

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df_clean, x='Spend', bins=20, kde=True)

plt.title('Customer Spending Distribution')
plt.xlabel('Amount Spent')
plt.ylabel('Number of Customers')

plt.show()
```

**Average Spending by Gender**

**Average Spending by Food Item**

```python
plt.figure(figsize=(9, 5))

sns.barplot(data=df_clean, x='Item', y='Spend')

plt.title('Average Spending by Food Item')
plt.xlabel('Food Item')
plt.ylabel('Average Spending')

plt.xticks(rotation=30)
plt.show()
```

Answer Questions Using the Data

```python
most_popular_item = df_clean['Item'].value_counts().idxmax()
item_count = df_clean['Item'].value_counts().max()
print("Most Popular Food Item?", most_popular_item)
print("Number of Records?", item_count)

print(50*"_")
most_common_city = df_clean['City'].value_counts().idxmax()
city_count = df_clean['City'].value_counts().max()
print("City with Most Records?", most_common_city)
print("Number of Records?", city_count)

print(50*"_")
gender_counts = df_clean['Gender'].value_counts()
print(gender_counts)
print("Gender with More Records:", gender_counts.idxmax())

print(50*"_")
average_spending = df_clean['Spend'].mean()
print("Average Spending?", round(average_spending, 2))

print(50*"_")
item_average_spending = df_clean.groupby('Item')['Spend'].mean()
print(item_average_spending.sort_values(ascending=False))
print("\nHighest Average Spending Item?",
      item_average_spending.idxmax())

print(50*"_")
frequency_counts = df_clean['Frequency'].value_counts()
print(frequency_counts)
print("\nMost Common Frequency?", frequency_counts.idxmax())
```
